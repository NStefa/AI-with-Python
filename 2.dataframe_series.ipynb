## 2. NumPy - fundament obliczeń numerycznych

**NumPy** (*Numerical Python*) to biblioteka, na której zbudowany jest cały dział data science w Pythonie - w tym Pandas, scikit-learn, TensorFlow i PyTorch. Jej sercem jest typ `ndarray`: tablica **jednorodna** (wszystkie elementy tego samego typu) i **o stałym rozmiarze**.

### Dlaczego nie wystarczy zwykła lista?

| | Lista Pythona | Tablica NumPy |
|---|---|---|
| Typy elementów | dowolne, mieszane | jeden wspólny (`int64`, `float64`, ...) |
| Pamięć | każdy element to osobny obiekt + wskaźnik | ciągły blok pamięci |
| Operacje matematyczne | trzeba pisać pętlę | **wektoryzacja** - działa na całej tablicy |
| Szybkość | wolno (interpreter Pythona) | szybko (skompilowany kod C) |
| Wielowymiarowość | listy list, niewygodne | naturalna, `.shape` |

````python
# Tablica z listy
a = np.array([1, 2, 3, 4, 5])
print("tablica:", a)
print("shape:", a.shape, "| dtype:", a.dtype)

# Tablica 2D (macierz)
m = np.array([[1, 2, 3],
              [4, 5, 6]])
print("\nmacierz:\n", m)
print("shape:", m.shape, "| ndim:", m.ndim)

# Gotowe generatory tablic
print("\narange:  ", np.arange(0, 10, 2))     # od 0 do 10 co 2 (bez ostatniej liczby)
print("linspace:", np.linspace(0, 1, 5))       # 5 równych punktów 0..1
print("zeros:   ", np.zeros(3)) # tablica 3 elementowa wypełniona zerami
````

### Wektoryzacja - serce NumPy

Najważniejsza cecha: **jedno działanie na całej tablicy naraz**, bez pętli. 

````python
ceny = np.array([100, 250, 399, 89])

# W czystym Pythonie trzeba pętli / listy składanej:
# z_vat = [c * 1.23 for c in ceny]

# W NumPy - jedno działanie na całej tablicy:
print("ceny z VAT:", (ceny * 1.23).round(2))

# Funkcje NumPy też działają element po elemencie:
print("pierwiastki:", np.sqrt(np.array([1, 4, 9, 16])))
````

> **CIEKAWOSTKA - dlaczego nie zwykła lista list?**
>
> Lista Pythona trzyma każdą liczbę jako osobny obiekt rozrzucony po pamięci, a operacje idą powolną pętlą interpretera. Tablica NumPy leży w jednym ciągłym bloku pamięci, a działania wykonują się w skompilowanym kodzie C. Zmierzmy różnicę na milionie liczb:

````python
lista = list(range(1_000_000))
tablica = np.arange(1_000_000)

print("Czysty Python (pętla po liście):")
%timeit [x * 2 for x in lista]

print("\nNumPy (wektoryzacja):")
%timeit tablica * 2
````

### Agregacje i oś (`axis`)

`sum`, `mean`, `std` liczą całość albo wzdłuż wybranej **osi**: `axis=0` w dół kolumn, `axis=1` wzdłuż wierszy. To samo pojęcie wróci w Pandas i przy redukcji macierzy w modelach.

````python
m = np.array([[1, 2, 3],
              [4, 5, 6]])
print("suma całości:     ", m.sum())
print("suma kolumn (0):  ", m.sum(axis=0)) 
print("suma wierszy (1): ", m.sum(axis=1))  
print("średnia:", m.mean(), "| odch. std:", m.std().round(2))
````

### Dodawanie i odejmowanie

Działają **element po elemencie** - macierze muszą mieć **ten sam kształt**.

````python
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[10, 20],
              [30, 40]])

print("A + B =\n", A + B)
print("\nA - B =\n", A - B)
````

### Mnożenie: element po elemencie (`*`) vs macierzowe (`@`)

To najważniejsze rozróżnienie i najczęstsza pomyłka:
- `*` mnoży **odpowiadające sobie elementy**,
- `@` to **prawdziwe mnożenie macierzy** (wiersz razy kolumna).

````python
print("A * B  (element po elemencie):\n", A * B)
print("\nA @ B  (mnożenie macierzy):\n", A @ B)
````

### Reguła wymiarów macierzy

Przy mnożeniu macierzy (nie tylko w Pythonie - ale też w matematyce) **wewnętrzne wymiary muszą się zgadzać**:

$$(n \times m) \; @ \; (m \times p) = (n \times p)$$

Liczba kolumn pierwszej = liczba wierszy drugiej.

````python
X = np.array([[1, 2, 3],
              [4, 5, 6]])       
W = np.array([[1, 0],
              [0, 1],
              [1, 1]])          

wynik = X @ W                   
# Czy to działanie wykona się poprawnie?
````

### Działania na macierzach robi każda warstwa sieci neuronowej

Warstwa sieci to: `dane @ wagi + bias` - mnożenie i dodawanie macierzy.

````python
# Sprawdź wymiary macierzy zanim wykonasz kod!
dane = np.array([[0.5, 0.2, 0.1]])      
wagi = np.array([[0.8],
                 [0.1],
                 [0.3]])               
bias = np.array([0.05])

wynik = dane @ wagi + bias
````

### Transpozycja (`.T`)

Zamienia wiersze z kolumnami. Pojawia się wszędzie w ML - m.in. po to, żeby dopasować kształty do mnożenia macierzy.

````python
print("X:\n", X, "-> kształt", X.shape)
print("\nX.T (transpozycja):\n", X.T, "-> kształt", X.T.shape)
````

### NumPy a Pandas - jak się mają do siebie?

- każda kolumna `DataFrame` jest wewnętrznie tablicą NumPy (`df["kolumna"].to_numpy()`),
- `NaN` (brakująca wartość) to `np.nan`,

## 3. Series - jednowymiarowa struktura danych

`Series` to podstawowy budulec Pandasa - jednowymiarowa tablica z **etykietami (indeksem)**. Możesz myśleć o niej jak o pojedynczej kolumnie arkusza kalkulacyjnego.


````python
# Tworzenie Series z listy
oceny = pd.Series([5, 4, 6, 3, 5])
print(oceny)
print(f"\nTyp: {type(oceny)}")
print(f"Typ danych: {oceny.dtype}")
````

````python
# Series z własnym indeksem
oceny = pd.Series(
    [5, 4, 6, 3, 5],
    index=["Ania", "Bartek", "Celina", "Darek", "Ewa"]
)
print(oceny)
print(f"\nŚrednia: {oceny.mean():.2f}")
print(f"Max:     {oceny.max()}")
print(f"Min:     {oceny.min()}")
````

````python
# Series ze słownika - klucze stają się indeksem
populacja = pd.Series({
    "Warszawa": 1_794_000,
    "Kraków":     780_000,
    "Wrocław":    643_000,
    "Poznań":     545_000,
    "Gdańsk":     470_000
})
print(populacja)
print(f"\nŁączna populacja: {populacja.sum():,}")
````

````python
# Dostęp do elementów Series
print("Warszawa:", populacja["Warszawa"])
print("Pierwsze 3 miasta:\n", populacja[:3])
print("\nMiasta powyżej 600 tys.:\n", populacja[populacja > 600_000])
````

### Najważniejsze atrybuty i metody Series

| Atrybut/Metoda | Opis |
|----------------|------|
| `.index` | indeks (etykiety) |
| `.values` | wartości jako tablica NumPy |
| `.dtype` | typ danych |
| `.shape` | kształt (liczba elementów,) |
| `.size` | liczba elementów |
| `.mean()` | średnia |
| `.sum()` | suma |
| `.min()` / `.max()` | minimum / maksimum |
| `.describe()` | podsumowanie statystyczne |
| `.value_counts()` | zliczenie unikalnych wartości |
| `.isnull()` | maska brakujących wartości |


````python
# value_counts() - bardzo przydatne przy eksploracji danych
oceny_klasy = pd.Series([5, 4, 5, 3, 6, 4, 5, 2, 4, 6, 5, 3])
print("Rozkład ocen:")
print(oceny_klasy.value_counts().sort_index())
print(f"\nOpis statystyczny:")
print(oceny_klasy.describe())
````

## 4. DataFrame - tabela danych

`DataFrame` to dwuwymiarowa tabela z wierszami i kolumnami - jak arkusz Excela lub tabela w bazie danych. Każda kolumna to osobny obiekt `Series`.


````python
# Tworzenie DataFrame ze słownika list
uczniowie = pd.DataFrame({
    "imie":      ["Ania", "Bartek", "Celina", "Darek", "Ewa"],
    "klasa":     ["2A",   "2A",     "2B",     "2B",    "2A"],
    "matematyka":[5,       4,        6,        3,       5],
    "fizyka":    [4,       5,        5,        4,       3],
    "informatyka":[6,      6,        5,        4,       6]
})

print(uczniowie)
````

````python
# Podstawowe atrybuty DataFrame
print("Kształt (wiersze, kolumny):", uczniowie.shape)
print("Liczba wierszy:", len(uczniowie))
print("Kolumny:", list(uczniowie.columns))
print("Indeks:", list(uczniowie.index))
print("\nTypy danych:")
print(uczniowie.dtypes)
````

````python
# Tworzenie DataFrame z listy słowników
produkty = pd.DataFrame([
    {"nazwa": "Laptop",    "cena": 3499.99, "kategoria": "Elektronika", "sztuki": 15},
    {"nazwa": "Mysz",      "cena":   89.99, "kategoria": "Elektronika", "sztuki": 42},
    {"nazwa": "Biurko",    "cena":  799.00, "kategoria": "Meble",       "sztuki":  8},
    {"nazwa": "Krzesło",   "cena":  499.00, "kategoria": "Meble",       "sztuki": 12},
    {"nazwa": "Słuchawki", "cena":  249.99, "kategoria": "Elektronika", "sztuki": 27},
    {"nazwa": "Lampa",     "cena":  149.00, "kategoria": "Meble",       "sztuki": 20},
])

print(produkty)
````

| Funkcja | Format | Rozszerzenie |

### Nasz zbiór danych: `2_student_habits_performance.csv`

Zbiór zawiera **1000 studentów** i 16 kolumn:

| Kolumna | Opis |
|---------|------|
| `student_id` | identyfikator studenta |
| `age` | wiek (17-24) |
| `gender` | płeć (`Female`, `Male`, `Other`) |
| `study_hours_per_day` | godziny nauki dziennie |
| `social_media_hours` | godziny w mediach społecznościowych |
| `netflix_hours` | godziny oglądania seriali |
| `part_time_job` | praca dorywcza (`Yes`/`No`) |
| `attendance_percentage` | frekwencja na zajęciach (%) |
| `sleep_hours` | godziny snu |

| `internet_quality` | jakość internetu (`Poor`, `Average`, `Good`) |

### Najważniejsze parametry `read_csv()`

| Parametr | Opis | Przykład |
|----------|------|---------|
| `sep` | separator kolumn (domyślnie `,`) | `sep=';'` |
| `encoding` | kodowanie znaków | `encoding='utf-8'` |
| `index_col` | kolumna jako indeks | `index_col='student_id'` |
| `usecols` | wczytaj tylko wybrane kolumny | `usecols=['age', 'exam_score']` |
| `nrows` | wczytaj tylko N wierszy | `nrows=100` |
| `skiprows` | pomiń N wierszy na początku | `skiprows=2` |
| `na_values` | dodatkowe wartości traktowane jako NaN | `na_values=['brak', 'N/A']` |
| `dtype` | wymuś typy danych kolumn | `dtype={'age': 'int8'}` |


````python
# Wczytanie z kolumną student_id jako indeksem
df = pd.read_csv("2_student_habits_performance.csv", index_col="student_id", encoding="utf-8")
print(df.head(3))
print(f"\nPierwsze indeksy: {list(df.index[:5])}")
````

````python
# Wczytanie tylko wybranych kolumn - oszczędza pamięć przy dużych plikach
df_mini = pd.read_csv(
    "2_student_habits_performance.csv",
    usecols=["student_id", "study_hours_per_day", "sleep_hours", "exam_score"]
)
print(df_mini.head())
print("\nZużycie pamięci (MB):")
print(f"  pełny zbiór:  {pd.read_csv("2_student_habits_performance.csv").memory_usage(deep=True).sum() / 1e6:.2f}")
print(f"  4 kolumny:    {df_mini.memory_usage(deep=True).sum() / 1e6:.2f}")
````
